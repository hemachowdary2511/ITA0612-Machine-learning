## Experiment 1 – FIND-S Algorithm

```python
data = [
    ['Sunny','Warm','Normal','Yes'],
    ['Sunny','Warm','High','Yes'],
    ['Rainy','Cold','High','No'],
    ['Sunny','Warm','High','Yes']
]

h = ['0'] * (len(data[0]) - 1)

for row in data:
    if row[-1] == 'Yes':
        if h[0] == '0':
            h = row[:-1]
        else:
            for i in range(len(h)):
                if h[i] != row[i]:
                    h[i] = '?'
    print(h)

print("Final Hypothesis:", h)
```

## Experiment 2 – Candidate Elimination Algorithm

```python
data=[
    ['Sunny','Warm','Yes'],
    ['Sunny','Cold','Yes'],
    ['Rainy','Warm','No'],
    ['Sunny','Warm','Yes']
]

S=data[0][:-1]
G=['?']*len(S)

for x in data:
    if x[-1]=='Yes':
        for i in range(len(S)):
            if S[i]!=x[i]:
                S[i]='?'
    else:
        for i in range(len(S)):
            if S[i]!=x[i]:
                G[i]=S[i]
    print("S=",S,"G=",G)

print("Final S=",S)
print("Final G=",G)
```

## Experiment 3 – Decision Tree (ID3) Algorithm

```python
from sklearn.tree import DecisionTreeClassifier

X=[[0,0],[0,1],[1,0],[1,1]]
y=['No','No','Yes','Yes']

clf=DecisionTreeClassifier(criterion='entropy')
clf.fit(X,y)

print(clf.predict([[1,0]]))
```

## Experiment 4 – Artificial Neural Network Using Backpropagation

## Experiment 6 – Naïve Bayes Classifier

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix,accuracy_score

X=[[1],[2],[3],[6],[7],[8]]
y=[0,0,0,1,1,1]

m=GaussianNB()
m.fit(X,y)
p=m.predict(X)

print(confusion_matrix(y,p))
print("Accuracy:",accuracy_score(y,p))
```

## Experiment 9 – Comparison of Linear and Polynomial Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

X=[[1],[2],[3],[4],[5]]
y=[1,4,9,16,25]

print("Linear:", LinearRegression().fit(X,y).predict(X))

X2=PolynomialFeatures(2).fit_transform(X)
print("Poly:", LinearRegression().fit(X2,y).predict(X2))
```
